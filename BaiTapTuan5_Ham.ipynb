W5A10

```python
def isomorphic(a, b):
    mapping = {}

    for i in range(len(a)):
        c1 = a[i]
        c2 = b[i]

        if c1 in mapping:
            if mapping[c1] != c2:
                return False
        else:
            if c2 in mapping.values():
                return False
            mapping[c1] = c2

    return True

a = input()
b = input()

print(isomorphic(a, b))
```
