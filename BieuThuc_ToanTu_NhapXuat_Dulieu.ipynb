W2A3

```python
def tong(a,b):
    return a+b
def hieu(a,b):
    return a-b
def tich(a,b):
    return a*b
def du(a,b):
    return a%b
def ket_qua_chia(a,b):
    return a/b
a,b=map(int,input("Nhap 2 so nguyen bat ky,cach nhau boi dau cach:").split( ))
print(f"Tong={tong(a,b)}")
print(f"Hieu={hieu(a,b)}")
print(f"Tich={tich(a,b)}")
print(f"Du={du(a,b)}")
print(f"Ket qua phep chia={ket_qua_chia(a,b):.2f}")
```

```python
kitu=input("Nhap 1 ki tu thuong bat ki:")
if len(kitu)>1:
    print("Vuot qua so luong ki tu")
else:    
    code_kitu=ord(kitu)
    print(f"Ma Unicode cua ki tu vua nhap la:{code_kitu}")
    if 97<=code_kitu<=122: print(f"Ki tu in hoa la:{chr(code_kitu-32)}")
    elif 65<=code_kitu<=90:print("Day da la ki tu in hoa")
    else: print("Ki tu nhap vao khong hop le")    
```
