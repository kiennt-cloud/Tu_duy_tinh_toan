W3A11

```python
def check(a,b,c):
    if a+b>c and a+c>b and b+c>a: return 1
    else: return 0
def khongtamgiac():
    print("Khong phai tam giac")
def tamgiac(a,b,c):
    if a==b==c:
        print("Tam giac deu")
    elif a==b!=c or a==c!=b or b==c!=a:
        print("Tam giac can")
    else:
        print("Tam giac thuong")
def checkcanh(a,b,c):
    if a>0 and b>0 and c>0: return 1
    else: return 0
a,b,c=map(int,input().split( ))
if checkcanh(a,b,c)==1:
    if check(a,b,c)==1:
        tamgiac(a,b,c)
    else:
        khongtamgiac()    
else:
    print("Do dai canh khong hop le")                                
```
