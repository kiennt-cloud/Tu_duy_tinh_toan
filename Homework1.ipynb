```python
def chia_400(n):
    if n%400==0: return 1
    else: return 0
def chia_4(n):    
    if n%4==0: return 1
    else: return 0
def chia_100(n):   
    if n%100==0: return 1
    else: return 0
year=int(input("Nhap so nam can xem:"))
if((chia_400(year)==1)or((chia_4(year)==1)and(chia_100(year)==0))): print("Day la nam nhuan")
else: print("Day khong phai la nam nhuan")     
```

```python
def transfer(n):
    match n:
        case 0:print("Khong")
        case 1:print("Mot")
        case 2:print("Hai")
        case 3:print("Ba")
        case 4:print("Bon")
        case 5:print("Nam")
        case 6:print("Sau")
        case 7:print("Bay")
        case 8:print("Tam")
        case 9:print("Chin")
        case _:print("Du lieu nhap vao khong hop le")
n=int(input("Nhap so nguyen bat ky tu 0 den 9:"))
transfer(n)        
```
