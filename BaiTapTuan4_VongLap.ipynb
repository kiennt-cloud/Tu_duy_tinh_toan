```python
import math
def check(a):
    count=0
    if a>2:
        for i in range (2,int(math.sqrt(a))+1):
            if a%i==0:
                count=1
    elif a==2: count=0            
    return count        
a,b=map(int,input().split( ))
if 0<a<=b:
    sum=0
    if a>=2:
        for i in range (a,b+1):
            if check(i)==0:
                sum+=i  
        print(sum)
    else:
        for i in range (2,b+1):
            if check(i)==0:
                sum+=i  
        print(sum)
```

```python
import math
def check_so_nguyen_to(a):
    check=1
    if a<=1:check=0
    elif a>2:
        for i in range(2,int(math.sqrt(a))+1):
            if a%i==0:
                check=0
                break
    if check==1:return 1        
    return 0
n=int(input())
uoc=[]
for i in range (2,n+1):
    if n%i==0:
        uoc.append(i)
l=len(uoc)
max=0
for i in range (l):
    if check_so_nguyen_to(int(uoc[i]))==1 and int(uoc[i])>max:
        max=uoc[i]
print(max)                
```

W4A13

```python
def check_than_thiet(a,b):
    tong_a=0
    tong_b=0
    for i in range(1,a):
        if a%i==0:
            tong_a+=i
    for i in range(1,b):
        if b%i==0:
            tong_b+=i        
    if tong_a==b and tong_b==a:
        return True
    return False
a,b=map(int,input().split( ))
if a>0 and b>0:
    print(check_than_thiet(a,b))
```
